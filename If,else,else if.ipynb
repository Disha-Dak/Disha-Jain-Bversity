```python
def is_palindrome(sequence):
    # Create a dictionary for base complement
    complement = {'A': 'T', 'T': 'A', 'C': 'G', 'G': 'C'}
    
    # Get the reverse complement of the sequence
    reverse_complement = ''.join(complement[base] for base in sequence[::-1])
    
    # Check if the sequence is equal to its reverse complement
    return sequence == reverse_complement

# Example usage
dna_sequence = "ATGCAT"
palindrome_result = is_palindrome(dna_sequence)

if palindrome_result:
    print(f"The sequence {dna_sequence} is palindromic.")
else:
    print(f"The sequence {dna_sequence} is not palindromic.")
```
