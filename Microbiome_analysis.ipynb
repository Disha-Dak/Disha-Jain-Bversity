```python
import matplotlib.pyplot as plt
import seaborn as sns


# Step 1: Calculate total counts for Sample_1
total_counts_sample_1 = df['Sample_1'].sum()

# Step 2: Calculate relative abundance for Sample_1
relative_abundance_sample_1 = df['Sample_1'] / total_counts_sample_1

# Step 3: Get the top 10 taxa by relative abundance
top_10_taxa = relative_abundance_sample_1.nlargest(10)


# Step 4: Plot the bar chart
plt.figure(figsize=(8, 6))
top_10_taxa.plot(kind='bar', color='skyblue')
plt.title('Top 10 Taxa in Sample_1 by Relative Abundance')
plt.xlabel('taxa ')
plt.ylabel('relative_abundance_sample_1')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.tight_layout()
```
