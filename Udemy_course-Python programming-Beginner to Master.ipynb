```python
#reversed number
number = int(input("Enter your number: "))
rev = 0

while number > 0:
    r = number % 10  # Get the last digit
    number = number // 10  # Remove the last digit
    rev = rev * 10 + r  # Build the reversed number
print("Reversed number:", rev)

```

```python
#palindrome
number = int(input("Enter your number: "))
m = number
rev = 0  # Initialize the reversed number

while number > 0:
    r = number % 10  # Get the last digit
    number = number // 10  # Remove the last digit
    rev = rev * 10 + r  # Build the reversed number

if m == rev:
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
start = int(input("Enter the starting number of the range: "))
end = int(input("Enter the ending number of the range: "))

print(f"Prime numbers between {start} and {end} are:")

for num in range(start, end + 1):
    if num > 1:  # Prime numbers are greater than 1
        count = 0
        for i in range(2, num):  # Check for factors
            if num % i == 0:  # If divisible, increment the count
                count += 1
                break  # No need to check further
        if count == 0:  # If no factors were found, it's a prime number
            print(num)
```

\D: Non digit


Character	Description	Example
.	Matches any single character	a.c matches abc, adc
^	Matches the start of the string	^hello matches hello world
$	Matches the end of the string	world$ matches hello world
*	Matches 0 or more repetitions	a* matches aaa, a
+	Matches 1 or more repetitions	a+ matches aaa, a
?	Matches 0 or 1 occurrence	a? matches a, or empty
{n,m}	Matches between n and m repetitions	a{2,3} matches aa, aaa
[]	Matches any character inside brackets	[abc] matches a, b, c
`	`	Matches either pattern (OR)
\	Escape special characters	\. matches .

#set operators
Union (`	`)**	Combines all unique elements from both sets.
Intersection (&)	Returns only the common elements between sets.	A & B	
Difference (-)	Returns elements in the first set but not in the second.	A - B
Symmetric Difference (^)	Returns elements that are in either set but not in both.	A ^ B
Subset (<=)	Checks if all elements of the first set are in the second set.	A <= B
Proper Subset (<)	Checks if all elements of the first set are in the second set and A is not equal to B.	A < B
Superset (>=)	Checks if all elements of the second set are in the first set.	A >= B
Proper Superset (>)	Checks if all elements of the second set are in the first set and A is not equal to B.	A > B

```python
#Longest and shortest meaning in a dictionary
dictionary = {
    "cat": "A small domesticated carnivorous mammal.",
    "hippopotamus": "A large, thick-skinned, herbivorous mammal.",
    "ant": "A small insect that lives in colonies.",
    "elephant": "A large mammal with a trunk.",
    "bat": "A nocturnal flying mammal.",
    "pneumonoultramicroscopicsilicovolcanoconiosis": "A supposed lung disease caused by silica dust.",
}
```

```python
# Find the longest and shortest words
longest_word = max(dictionary, key=len)
shortest_word = min(dictionary, key=len)

# Display results
print(f"Longest Word: {longest_word}, Meaning: {dictionary[longest_word]}")
print(f"Shortest Word: {shortest_word}, Meaning: {dictionary[shortest_word]}")
```

```python
#Generator (nextupon) they remain in the same state
def days():
    L = ["Sunday" , "Monday" , "Tuesday " , "Wednesday " , "Thursday", "Friday", "Saturday "]
    i = 0   #iterating 


    while True:
        x = L[i]
        i = (i+1)%7
        yield x
    
d= days()
print(next(d))
print(next(d))
print(next(d))
print(next(d))
print(next(d))
print(next(d))
    
```

```python
def max3(n1, n2, n3):
    if n1 >= n2 and n1 >= n3:
        return n1
    elif n2 >= n1 and n2 >= n3:
        return n2
    else:
        return n3

print(max3(5, 10, 3))  # Output: 10
```
